```python
from sklearn.tree import DecisionTreeClassifier

feature  = [[150,1],
           [120,0],
           [144,2],]

labels = ["apple","banana","mango"]

clf = DecisionTreeClassifier()
clf = clf.fit(feature,labels)
prediction = clf.predict([[150, 1]])

print(f"The name of fruit is : {prediction[0]}")

```
