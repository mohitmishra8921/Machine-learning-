```python
import pandas as pd

df = pd.read_csv('housing.csv')

# Clean column names (strip spaces, make lowercase to match easily)
df.columns = df.columns.str.strip()

# Define your desired columns (checking common variations)
desired_columns = [
    "Households", "Population", "House_median_age", "Total_rooms", 
    "Total_bedrooms", "Longitude", "Latitude", "Median_income", 
    "Median_house_value", "Ocean_proximity"
]


# This maps columns case-insensitively so it won't crash
column_mapping = {col.upper(): col for col in df.columns}
matched_cols = []
for col in desired_columns:
    # Look for exact or lowercase match
    if col in df.columns:
        matched_cols.append(col)
    elif col.upper() in column_mapping:
        matched_cols.append(column_mapping[col.upper()])

housing_data = df[matched_cols].copy()
housing_data
```

```python
 housing_data.columns = [
    "Households", "Population",  "Total_rooms", 
    "Total_bedrooms", "Longitude", "Latitude", "Median_income", 
    "Median_house_value", "Ocean_proximity"
].copy()
housing_data
```

```python
housing_data
```

```python
df2 = housing_data.sort_values(["Households","Median_income"],ascending=True).copy()
df2
```

```python
df2.reset_index(drop=True,inplace= True)
df2
```

```python
Final_data = df2.to_csv("Cleaned_Housing_Data.csv")
Final_data
```
